# 01 - Tokenizer basics
Encode text into token IDs, decode them back, and compare tokenizers using `tiktoken` (runs offline after first download, no API key needed).

```python
import tiktoken

text = "Generative AI creates new text, images and code."
enc = tiktoken.get_encoding("cl100k_base")

token_ids = enc.encode(text)
print("Token IDs:", token_ids)
print("Number of tokens:", len(token_ids))
```

```python
# See each token as text
for tid in token_ids:
    print(tid, repr(enc.decode([tid])))
```

```python
decoded = enc.decode(token_ids)
print("Decoded:", decoded)
print("Round-trip OK:", decoded == text)
```

```python
# Compare encodings: same text, different token counts
for name in ["gpt2", "cl100k_base", "o200k_base"]:
    e = tiktoken.get_encoding(name)
    ids = e.encode(text)
    print(f"{name:12} -> {len(ids):3} tokens  {ids}")
```

**Explanation:** a tokenizer splits text into sub-word pieces and maps each to an integer ID; the model only sees these IDs. Decoding reverses the mapping. Newer encodings (larger vocabularies) usually need fewer tokens for the same text, which lowers cost and fits more into the context window.
